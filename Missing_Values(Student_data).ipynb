```python
import pandas as pd

df = pd.read_csv("students_data.csv", parse_dates=['enrolled_date'])
df
```

```python
new_df=df.fillna(0)
new_df
```

```python
new_df=df.fillna({
    'age':'no age',
    'math_score':0,
    'english_score':0
})
new_df
```

```python
new_df=df.ffill()
new_df
```

```python
new_df=df.bfill()
new_df
```

```python
df = df.interpolate()
new_df
```

```python
df = df.dropna()
new_df
```

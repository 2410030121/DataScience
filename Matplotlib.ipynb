```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.figure(figsize=(4,3))
plt.plot(x, y, color='r', marker='o', linestyle='dashed',linewidth=2, markersize=12 ) # line chart generation
plt.title("Line Chart Generation") #Title generaiion
plt.xlabel("x-Axis")
plt.ylabel("y-Axis")
plt.show()    # display chart
```

```python
x=['a','b','c','d','e']
y=[10,20,30,40,50]
plt.figure(figsize=(4,3))
plt.bar(x,y)
plt.title("Bar Graph Generation") #Title generaiion
plt.xlabel("x-Axis")
plt.ylabel("y-Axis")
plt.show()
```

```python
import random
data = [random.randint(1, 20) for _ in range(500)]
plt.hist(data, bins=20) # histogram chart
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

```python
#Pie chart is a relationship betweeen a data
categories = ['A','B','C','D','E']
sales = [10, 20, 55, 35, 45]
plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)
plt.title("Pie Chart Example")
plt.show()
```

```python
#Scattter plot is a dot-dot drawing
#Syntax is .scatter()
x = [1,2,3,4,5]
y = [10,20,30,40,50]
plt.scatter(x,y)
plt.xlabel("x-Axis")
plt.ylabel("y-Axis")
plt.title("Scatter Plot Example")
plt.show()
```
